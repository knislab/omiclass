# Publication plot regeneration

Run the pipeline once with `PDF: true`. The production `PlottingFramework` then regenerates figures from `pdf_plots/base_files` without loading models, retraining, or parsing the HTML report.

```python
from pathlib import Path
from omiclass.plotting.framework import PlottingFramework

run_dir = Path("/path/to/pipeline-run")
plots = PlottingFramework(run_dir)
```

Each plot family can be generated independently.

```python
# Classification
plots.plot_confusion_matrices()
plots.plot_pr_curve()
plots.plot_roc_curve()

# Batch-correction diagnostics
plots.plot_pc_metadata_heatmaps()
plots.plot_pca_scatters()
```

```python
# Or regenerate every available publication plot in one call.
generated_paths = plots.generate_all()
generated_paths
```
